```python
import torch
import torch.nn as nn
```

```python
# input
torch.manual_seed(42) # to get the same output every run
inputs = torch.tensor([1.0, 2.0, 3.0])
```

```python
neuron = nn.Linear(in_features=3, out_features=1) # Linear Activation Function
```

```python
output = neuron(inputs)
```

```python
# verify - using formula
neuron.weight @ inputs + neuron.bias
```
