```python
class Student1: 
    def __init__(self, name, marks): 
        self.name = name 
        self.marks = marks 
    
    def display(self):# Instance method 
        print(f"Name: {self.name}, Marks: {self.marks}, School: {Student1.school_name}")
        
    school_name = "ABC School" 
    
    @classmethod 
    def change_school(cls, new_name): 
        cls.school_name = new_name
        
    @staticmethod
    def add(a,b):  #normal functions belongs to the class
        return a+b 
        

# Creating objects (instances)
stu1 = Student1("Debashish", 10)
stu2 = Student1("Antor", 15)

stu1.display()
stu2.display()

# Changing school name using class method
Student1.change_school("XYZ School")

stu1.display()
stu2.display()
```

<h1> Encapsulation </h1>

```python
class Student2:
    def __init__(self, name, gpa, marks):
        self.name = name # public variable
        self._gpa = gpa # protected variable (can be accessed, not recommended)
        self.__marks = marks # private variable (cannot be accesed directly from outside, but can be accessed using _ClassName_variable format)
        
s = Student2("Debashish", 3.8, 100)

#access and print the variable
print("Public:", s.name)
print("Protected:", s._gpa)
print("Private:", s._Student2__marks) #name-mangling
```

```python
class Employee: 
    def __init__(self, salary): 
        self.__salary = salary # private 
        
    def get_salary(self): # getter 
        return self.__salary 
        
    def set_salary(self, new_salary): # setter 
        self.__salary = new_salary 
            
e = Employee(50000) 
print(e.get_salary()) 
e.set_salary(60000)
print(e.get_salary()) 
```

<h1>Inheritance</h1>

```python
# single level inheritance
class Employee:
    start_time = "10am"
    end_time = "6pm"

class Teacher(Employee):
    def __init__(self, subject):
        self.subject = subject
        
        
class AdminStaff(Employee):
    def __init__(self, role):
        self.role = role

t1 = Teacher("Math")
s1 = AdminStaff("Manager")

print(t1.subject, t1.start_time, t1.end_time, s1.role)

# private variable cannot be accessed in subclass, 
# but protected varibale can be accessed in subclass
```

```python
# multilevel level inheritance
class Employee:
    start_time = "10am"
    end_time = "6pm"
        
class AdminStaff(Employee):
    def __init__(self, role):
        self.role = role

class Accountant(AdminStaff):
    def __init__(self, salary, role):
        super().__init__(role) #invoke the constructor of parent class
        self.salary = salary


acc1 = Accountant(25000, "CA")

print(acc1.role, acc1.salary, acc1.start_time, acc1.end_time)
```

```python
# multiple inheritance 
# TA, who is a teacher and also a student

class Teacher:
    def __init__(self, salary):
        self.salary = salary
        
class Student:
    def __init__(self, gpa):
        self.gpa = gpa
        
class TA(Teacher, Student):
    def __init__(self, salary, gpa, name):
        super().__init__(salary)
        Student.__init__(self, gpa) 
        # using super method does not need self parameter, but invoking the parent class directly, needs self
        self.name = name
        
ta1 = TA(15000, 9.3, "Debashish")

print(ta1.name, ta1.salary, ta1.gpa)
```

<h1>Abstraction </h1>

```python
# Abstract class = blueprint for other classes
# abc module = abstraction based classes module

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound():
        pass # represents null value, which means this method is not yet implemented
    
class Lion(Animal):
    def make_sound(self):
        print("roar")
        
        
class Cat(Animal):
    def make_sound(self):
        print("meow")
        
"""lion = Lion()
lion.make_sound()

cat = Cat()
cat.make_sound()"""

def make_it_speak(entity):
    entity.make_sound()# doesn’t care about type
    
l = Lion()
c = Cat()
for e in[l, c]:
    make_it_speak(e)
```

<h3>Mini Chat System</h3>

```python
# -------------------------------
# Message Class
# -------------------------------
class Message:
    message_counter = 1   # simple counter

    def __init__(self, sender, content):
        self.sender = sender
        self.content = content
        self.id = Message.message_counter
        Message.message_counter += 1

    def __str__(self):
        return f"({self.id}) {self.sender.username}: {self.content}"


# -------------------------------
# User Class
# -------------------------------
class User:
    def __init__(self, username):
        self.username = username
        self.chatroom = None

    def join_chatroom(self, chatroom):
        if self.chatroom:
            print(f"{self.username} is already in a chatroom.")
        else:
            chatroom.add_user(self)
            self.chatroom = chatroom
            print(f"{self.username} joined {chatroom.name}")

    def leave_chatroom(self):
        if not self.chatroom:
            print(f"{self.username} is not in any chatroom.")
        else:
            self.chatroom.remove_user(self)
            print(f"{self.username} left {self.chatroom.name}")
            self.chatroom = None

    def send_message(self, content):
        if not self.chatroom:
            print(f"{self.username} cannot send a message (not in a chatroom).")
        else:
            self.chatroom.broadcast(self, content)


# -------------------------------
# ChatRoom Class
# -------------------------------
class ChatRoom:
    def __init__(self, name):
        self.name = name
        self.users = []
        self.messages = []

    def add_user(self, user):
        self.users.append(user)

    def remove_user(self, user):
        self.users.remove(user)

    def broadcast(self, sender, content):
        message = Message(sender, content)
        self.messages.append(message)
        print(message)   # Show message to all users

    def show_chat_history(self):
        print(f"\nChat History of {self.name}:")
        for msg in self.messages:
            print(msg)
        print()


# ---------------------------------------
# Example Usage
# ---------------------------------------
if __name__ == "__main__":
    room = ChatRoom("Python Lounge")

    u1 = User("Alice")
    u2 = User("Bob")
    u3 = User("Charlie")

    u1.join_chatroom(room)
    u2.join_chatroom(room)

    u1.send_message("Hello everyone!")
    u2.send_message("Hi Alice!")

    u3.join_chatroom(room)
```
