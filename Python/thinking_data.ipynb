```python
import json
```

```python
# Load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)

        return data
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
```

```python
# Clean & structure the data
def clean_data(data):
    text_to_num = {"one": 1, "two" : 2, "three" : 3, "four" : 4, "five" : 5}
    cleaned_data = []
    unique_users = set()
    
    for user in data:
        #clean ratings - consistent data type
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating

        #handle missing values
        raw_age = user.get("age") #get method returns None, if it's value not found
        if(raw_age == None):
            user["age"] = None

        #Deduplication
        if(user["name"].strip() in unique_users):
            continue
            
        unique_users.add(user["name"])
        cleaned_data.append(user)
        
    return cleaned_data
```

```python
data = clean_data(data)
print(data)
```

```python
# get meaningful insights
def get_insights(data):

    #avg rating
    total_rating = 0
    for user in data:
        total_rating += float(user["rating"])
    print(f"average rating: {total_rating/len(data)}")

    #percentage of users with poor rating
    poor_ratings = 0
    for user in data:
        if(float(user["rating"]) < 3):
            poor_ratings += 1
            
    print(f"% of user with poor rating = {poor_ratings/len(data) * 100}%")
            
    
```

```python
get_insights(data)
```

```python
# Build the recommendation feature (Rule based)
def get_recommendations(data):
    recommendations = []

    for user in data:
        curr_recom = {} #this dict contains user name and recommended brand
        curr_recom["name"] = user["name"]
        if(float(user["rating"]) >= 4):
            curr_recom["brand"] = "Apple"
        else:
            curr_recom["brand"] = "Samsung"

        recommendations.append(curr_recom)
    return recommendations
```

```python
get_recommendations(data)
```
