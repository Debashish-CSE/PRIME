```python
import numpy as np
```

```python
# Linear Regression

class LinearRegression:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias = None # theta_0
        self.weights = None #theta_i
        self.lr = learning_rate
        self.n_iter = n_iter
        
        
    def fit(self, X, y): # X_train, y_train
        m,n = X.shape  # number of samples(row), number of features(col)
        
        # step 1 (Initialize params)
        self.bias = 0
        self.weights = np.zeros(n)    # n is the number of features, which indicates the number of weights(theta). array of 0
        
        # Gradient Descent
        for i in range(self.n_iter):
            # step 2 (y_pred = theta_0 + x1*theta_1 +....+ xn*theta_n)
            y_pred = self.bias + np.dot(X, self.weights)
            
            # step 3 (calc gradient)
            db = (1/m) * np.sum(y_pred - y)
            dw = (1/m) * np.dot(X.T, (y_pred - y)) 
            
            # step 4 (convergence theorem - params update)
            self.bias = self.bias - self.lr * db
            self.weights = self.weights - self.lr * dw
        
                
    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights) 
        return y_pred
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

print(model.bias) # 0 in this case
print(model.weights) # 2 in this case
```

```python
# Linear regression with OLS (Ordinary Least Squares)

class LinearRegression:
    def __init__(self):
        self.bias = None # theta_0
        self.weights = None #theta_i
      
    def fit(self, X, y):
        m,n = X.shape
        
        X_b = np.c_[np.ones((m, 1)), X] # concatenate extra column of 1 at the beginning of X, to calculate bias(theta_0)
        
        # normal equation
        # theta = np.dot( np.linalg.inv(np.dot(X_b.T, X_b)), np.dot(X_b.T, y))
        theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y  # @ = dot product
        
        self.bias = theta[0]
        self.weights = theta[1:]
    
    
    
        
    def predict(self, X):
        y_pred = self.bias + np.dot(X, self.weights) 
        return y_pred
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
print(y_pred)

print(model.bias) # 0 in this case
print(model.weights) # 2 in this case
```
