## SmartCart - E-commerce Customer Segmentation System

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("smartcart_customers.csv")
```

```python
df.shape
```

```python
df.isnull().sum()
```

### Data Preprocessing

##### Handle Missing Values

```python
df["Income"] = df["Income"].fillna(df["Income"].median())
```

```python
df.isnull().sum()
```

### Feature Engineering

```python
# Age
df["Age"] = 2026-df["Year_Birth"]
```

```python
# Customer Tenure Days
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], dayfirst=True)

reference_date = df["Dt_Customer"].max()

df["Customer_Tenure_Days"] = (reference_date - df["Dt_Customer"]).dt.days
```

```python
# Total Spending
df["Total_Spending"] = df["MntFishProducts"] + df["MntFruits"] + df["MntGoldProds"] + df["MntMeatProducts"] + df["MntSweetProducts"] + df["MntWines"]
```

```python
# Total Children
df["Total_Children"] = df["Kidhome"] + df["Teenhome"]
```

```python
# Education
df["Education"].value_counts()

# Undergraduate, Graduate, Postgraduate
df["Education"] = df["Education"].replace({
    "Basic":"Undergraduate", "2n Cycle": "Undergraduate",
    "Graduation": "Graduate",
    "Master": "Postgraduate", "PhD": "Postgraduate"
})
```

```python
# Marital Status 

df["Marital_Status"].value_counts()

# Living_With - Partner, Alone
df["Living_With"] = df["Marital_Status"].replace({
    "Married":"Partner", "Together": "Partner",
    "Single": "Alone", "Divorced": "Alone",
    "Widow": "Alone", "Absurd": "Alone", "YOLO": "Alone"
})
```

#### Drop Columns

```python
cols = ["ID", "Year_Birth", "Marital_Status", "Kidhome", "Teenhome", "Dt_Customer"]
spending_cols = ["MntWines", "MntFruits", "MntMeatProducts", "MntFishProducts", "MntSweetProducts", "MntGoldProds"]

cols_to_drop = cols + spending_cols

df_cleaned = df.drop(columns=cols_to_drop)
```

```python
df_cleaned.shape
```

```python
df_cleaned.head()
```

### Outliers

```python
cols =["Income", "Recency", "Response", "Age", "Total_Spending", "Total_Children"]

# relative plots of some features - pair plots
sns.pairplot(df_cleaned[cols])
```

```python
# Remove outliers

print("data size with outliers:", len(df))

df_cleaned = df_cleaned[(df_cleaned["Age"] < 90)]
df_cleaned = df_cleaned[(df_cleaned["Income"] < 600_000)]

print("data size without outliers:", len(df_cleaned))
```

### Encoding

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder() # we do not use drop_first="True" for clustering problems

cat_cols = ["Education", "Living_With"]

enc_cols = ohe.fit_transform(df_cleaned[cat_cols])
```

```python
enc_df = pd.DataFrame(enc_cols.toarray(), columns=ohe.get_feature_names_out(cat_cols), index=df_cleaned.index)
```

```python
df_encoded = pd.concat([df_cleaned.drop(columns=cat_cols), enc_df], axis=1)
```

```python
df_encoded.head()
```

### Scaling

```python
X = df_encoded
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

### Visualize

```python
# 2D 
from sklearn.decomposition import PCA

pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)
```

```python
# 3D Plot
fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:,0], X_pca[:,1], X_pca[:,2])
ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
ax.set_title("3D projection")
```

```python
pca.explained_variance_ratio_ # only 40% so not enough, this is only for visualization
```

### Analyze K value
#### 1. Elbow Method

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator

wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
    
```

```python
knee = KneeLocator(range(1,11), wcss, curve="convex", direction= "decreasing")
optimal_k = knee.elbow
```

```python
print("Optimal K:", optimal_k)
```

```python
# Plot

plt.plot(range(1,11), wcss, marker='o')
```

#### 2. Silhouette Score

```python
from sklearn.metrics import silhouette_score

scores = []
for k in range(2,11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_pca)
    score = silhouette_score(X_pca, labels)
    
    scores.append(score)
```

```python
# Plot
plt.plot(range(2,11), scores, marker="o")
plt.xlabel("K")
plt.ylabel("Silhouette score")
```

```python
# Combined plot for elbow and ss

k_range = range(2,11)

fig, ax1 = plt.subplots(figsize=(8,6))

ax1.plot(k_range, wcss[:len(k_range)], marker='o', color="blue")
ax1.set_xlabel("K")
ax1.set_ylabel("WCSS")

ax2 = ax1.twinx()
ax2.plot(k_range, scores[:len(k_range)], marker='x', color="red", linestyle="--")
ax2.set_ylabel("Silhouette Score")
```

##### Intersection point is around 4, so K=4 is chosen

### Clustering

```python
# K_Means

kmeans = KMeans(n_clusters=4, random_state=42)
labels_kmeans = kmeans.fit_predict(X_pca)
```

```python
fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:,0], X_pca[:,1], X_pca[:,2], c=labels_kmeans)
```

```python
# Agglomerative Clustering
from sklearn.cluster import AgglomerativeClustering

agg_clf = AgglomerativeClustering(n_clusters=4, linkage="ward")
labels_agg = agg_clf.fit_predict(X_pca)
```

```python
fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:,0], X_pca[:,1], X_pca[:,2], c=labels_agg)
```

##### Agglomerative Clustering gives us better clustering, so we use this for characterization and profiling

### Characterization of Clusters

```python
X["cluster"] = labels_agg
```
